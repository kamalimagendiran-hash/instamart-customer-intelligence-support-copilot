Customer Cleaning

```python
customers_clean = customers.copy()

# Standardize city names
customers_clean["city"] = (
    customers_clean["city"]
    .str.strip()
    .str.title()
)

# Standardize gender
gender_map = {
    "MALE": "male",
    "M": "male",
    "FEMALE": "female",
    "F": "female"
}

customers_clean["gender"] = (
    customers_clean["gender"]
    .str.strip()
    .str.upper()
    .map(gender_map)
    .fillna("unknown")
)

print(customers_clean["city"].value_counts())
print(customers_clean["gender"].value_counts())
```

Age cleaning

```python
# Treat ages outside 10–100 as missing
customers_clean.loc[
    (customers_clean["age"] < 10) |
    (customers_clean["age"] > 100),
    "age"
] = pd.NA

# Calculate city-wise median and fill missing ages
city_median = (
    customers_clean.groupby("city")["age"]
    .transform("median")
)

customers_clean["age"] = (
    customers_clean["age"].fillna(city_median)
)

print("Missing ages:", customers_clean["age"].isna().sum())
print("Minimum age:", customers_clean["age"].min())
print("Maximum age:", customers_clean["age"].max())
```

Signup Date Conversion to DateTime

```python
customers_clean["signup_date"] = pd.to_datetime(
    customers_clean["signup_date"],
    errors="coerce"
)

print(customers_clean["signup_date"].dtype)
print("Missing signup dates:", customers_clean["signup_date"].isna().sum())
```

# Customers Dataset – Data Audit Findings

## 1. Dataset Overview
- Total Records: 1,200
- Total Columns: 5
- Columns: customer_id, city, signup_date, age, gender

## 2. Issues Identified and Proposed Solutions

### Issue 1: Missing Age Values
**Finding:**
- 61 missing values in the age column.

**Proposed Solution:**
- Investigate the missing values.
- Do not automatically replace them with zero.
- Use median imputation if appropriate after checking the age distribution.

**Reason:**
- Missing age values can affect customer segmentation and analysis.

### Issue 2: Invalid Age Values
**Finding:**
- 5 suspicious age values were identified: -4, 150, 200, 280 and 999.

**Proposed Solution:**
- Treat these values as invalid.
- Convert them to missing values after validating the business rules.
- Consider median imputation or retain them as missing if imputation is not appropriate.

**Reason:**
- Invalid age values can distort statistical analysis and ML model predictions.

### Issue 3: Missing Gender Values
**Finding:**
- 30 missing values in the gender column.

**Proposed Solution:**
- Preserve missing values as unknown.
- Do not assume or randomly assign a customer's gender.

**Reason:**
- Incorrect gender assignment can create misleading customer insights.

### Issue 4: Inconsistent Gender Formatting
**Finding:**
- Gender values contain inconsistent formats:
  - Female, FEMALE, female, F, f
  - Male, MALE, male, M, m

**Proposed Solution:**
- Standardize female variations as Female.
- Standardize male variations as Male.

**Reason:**
- Consistent categories are required for accurate grouping and analysis.

### Issue 5: Inconsistent City Formatting
**Finding:**
- City names contain uppercase, lowercase and leading-space variations.
- Example: Chennai, chennai, CHENNAI and " Chennai".

**Proposed Solution:**
- Remove leading and trailing spaces.
- Standardize city names using a consistent format.

**Reason:**
- Inconsistent city names can create duplicate categories in customer analysis.

### Issue 6: Signup Date Data Type
**Finding:**
- signup_date is stored as an object/string data type.

**Proposed Solution:**
- Convert signup_date into a proper datetime data type.
- Validate dates before conversion.

**Reason:**
- Datetime format is required for time-based customer analysis.

## 3. Summary

The Customers dataset contains missing values, invalid age values, inconsistent categorical formatting and a date data type issue.

These issues will be addressed during the data-cleaning stage after completing the audit of all datasets.

**Proposed Solution:**
- Investigate tickets with missing order IDs.
- Check whether the ticket can be linked to an order using reliable source information.
- If no valid order reference exists, retain the value as missing.

**Reason:**
- Order references are useful for connecting support tickets with order records.

### Issue 2: Missing Ticket Text
**Finding:**
- 4 missing values in the text column.

**Proposed Solution:**
- Check whether the original ticket message can be recovered.
- If recovery is not possible, retain it as missing.
- Do not generate or assume customer messages.

**Reason:**
- Ticket text is important for text classification, NLP and support-agent responses.

### Issue 3: Category Value – none
**Finding:**
- 4 records have the literal category value "none".

**Proposed Solution:**
- Investigate these records to understand whether "none" represents an unknown category or an intentionally unclassified ticket.
- Map it to an appropriate category only after validation.
- Otherwise, retain it as an unknown/unclassified value.

**Reason:**
- Incorrect category assignment can affect ticket classification model training and evaluation.

### Issue 4: Repeated Customer and Order References
**Finding:**
- 164 tickets contain 157 unique customer IDs.
- 156 non-missing order IDs contain 150 unique values.
- Some customers and orders therefore appear in multiple tickets.

**Proposed Solution:**
- Retain these records unless they are confirmed to be duplicate tickets.
- Do not remove records just because customer IDs or order IDs repeat.

**Reason:**
- A customer or order can legitimately have multiple support tickets.

### Issue 5: Repeated Ticket Text
**Finding:**
- Some ticket messages are repeated.

**Proposed Solution:**
- Inspect repeated messages before taking action.
- Retain legitimate repeated complaints.
- Remove records only if they are confirmed to be duplicate tickets.

**Reason:**
- Repeated messages do not automatically mean duplicate records.
