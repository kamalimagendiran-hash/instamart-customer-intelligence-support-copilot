```python
customers = pd.read_csv("../data/raw/customers.csv")
```

```python
print("Customers:", customers.shape)
```

```python
customers_clean = customers.copy()


print("Working copies created successfully!")
```

```python
invalid_ages = [-4, 150, 200, 280, 999]

print(
    customers_clean[
        customers_clean["age"].isin(invalid_ages)
    ]
)
```

```python
print(customers_clean["city"].nunique())

print(customers_clean["city"].unique())
```

```python
print(customers_clean["signup_date"].dtype)

print(customers_clean["signup_date"].nunique())

print(customers_clean["signup_date"].isnull().sum())
```

```python
customers_clean["signup_date"] = pd.to_datetime(
    customers_clean["signup_date"],
    errors="coerce"
)
```

```python
print(customers_clean["signup_date"].dtype)

print(customers_clean["signup_date"].isnull().sum())
```

```python
print("CUSTOMERS FINAL AUDIT")

print("Shape:", customers_clean.shape)

print("\nMissing Values:")
print(customers_clean.isnull().sum())

print("\nDuplicates:")
print(customers_clean.duplicated().sum())

print("\nData Types:")
print(customers_clean.dtypes)
```

AGE VALIDATION

```python
invalid_ages = [-4, 150, 200, 280, 999]

customers_clean.loc[
    customers_clean["age"].isin(invalid_ages),
    "age"
] = np.nan
```

```python
print(customers_clean["age"].isnull().sum())
```

```python
median_age = customers_clean["age"].median()

print(median_age)
```

```python
customers_clean["age"] = customers_clean["age"].fillna(median_age)
```

```python
print("Missing:", customers_clean["age"].isnull().sum())
print("Min:", customers_clean["age"].min())
print("Max:", customers_clean["age"].max())
print("Median:", customers_clean["age"].median())
```

Gender validation

```python
customers_clean["gender"] = (
    customers_clean["gender"]
    .str.strip()
    .str.lower()
)
```

```python
gender_mapping = {
    "female": "Female",
    "f": "Female",
    "male": "Male",
    "m": "Male"
}

customers_clean["gender"] = customers_clean["gender"].map(gender_mapping)
```

```python
print(customers_clean["gender"].value_counts(dropna=False))
```

City Cleaning

```python
customers_clean["city"] = (
    customers_clean["city"]
    .str.strip()
    .str.title()
)
```

```python
print(customers_clean["city"].nunique())

print(customers_clean["city"].value_counts())
```

final validation 

```python
print("FINAL CUSTOMERS DATA")

print("\nShape:")
print(customers_clean.shape)

print("\nMissing Values:")
print(customers_clean.isnull().sum())

print("\nDuplicates:")
print(customers_clean.duplicated().sum())

print("\nData Types:")
print(customers_clean.dtypes)

print("\nUnique Cities:")
print(customers_clean["city"].unique())

print("\nUnique Genders:")
print(customers_clean["gender"].unique())
```

```python
customers_clean.to_csv(
    "customers_cleaned.csv",
    index=False
)
```
